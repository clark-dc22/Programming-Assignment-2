```python
import numpy as np
```

# Problem A

```python
np.random.seed(2112) #Setting a seed for the 5x5 array
X = np.random.randint(10, 101, size=(5,5))

X_mean = X.mean() #Solving for mean
X_std = X.std() #Solving for standard deviation
X_normalized = (X - X_mean) / X_std #Solving for normalization of the array

#Displaying array, mean, standard deviation, and normalized array
print("Array:")
print(X)
print("Normalized Array")
print(X_normalized)
print("Mean of X_normalize (should be 0):", X_normalized.mean())
print("Standard Deviation of X_normalize (should be 1):", X_normalized.std())

#Saving the array into a file
np.save('X_normalized.npy', X_normalized)
```

# Problem B

```python
C = (np.arange(1, 101) ** 3).reshape(10, 10) #Creating the first 100 positive integers and cubing them and turning them into a 10x10 array.

divide_by_4 = C[C % 4 == 0] #Specifying cubed integers only divisible by 4 

#Printing array, number of elements, first element, and last element
print("Shape of array C:", C.shape)
print("Array divide_by_4:")
print(divide_by_4)
print("Number of selected elements:", divide_by_4.size)
print("First element:", divide_by_4[0])
print("Last element:", divide_by_4[-1])

#Saving array into a file
np.save('div_by_4.npy', divide_by_4)
```

# Problem C

```python
S = (np.arange(1, 37) ** 2).reshape(6, 6) #Creating a 6x6 array with the square of the first 36 positive integers.

S_mean = S.mean() #Mean of array

above_mean = S[S > S_mean] #Integers that are only above the mean

#Printing array, mean, integers above the mean, number of selected integers, first element, and last element.
print("Array S:")
print(S)
print("\nMean of S (S_mean):", S_mean)
print("\nArray above_mean:")
print(above_mean)
print("\nNumber of selected elements:", above_mean.size)
print("First element:", above_mean[0])
print("Last element:", above_mean[-1])

# Saving array into a file
np.save('above_mean.npy', above_mean)
```
